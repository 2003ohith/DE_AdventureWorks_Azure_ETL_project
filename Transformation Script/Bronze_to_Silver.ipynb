## ***Silver Transformation Script***

###***Storage Container Path***

```python
brz_path="abfss://bronze@stgactadventureworks.dfs.core.windows.net/"
slv_path="abfss://silver@stgactadventureworks.dfs.core.windows.net/"
gld_path="abfss://gold@stgactadventureworks.dfs.core.windows.net/"
```

###***Schema Path***

```python
Silver_schema_path=f"Adventureworks.Silver_schema"
```

###***Calendar***

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

Raw_data_path = f"{brz_path}AdventureWorks_Calendar"

df_slv_calendar = spark.read.format("csv").option("header", True).option("inferSchema", True).load(Raw_data_path)
df_slv_calendar = df_slv_calendar.withColumn("Year", year(col("Date"))).withColumn("Quarter", concat(lit("Q"), quarter(col("Date")))).withColumn('Month_name',date_format("Date", "MMMM")).withColumn('Month',month(col("Date"))).withColumn("_source_file", col("_metadata.file_path")).withColumn("injested_at",current_timestamp())

df_slv_calendar.write.mode('overwrite').format("delta").option("overwriteSchema", "true").option("path",f"{slv_path}AdventureWorks_Calendar").saveAsTable(f"{Silver_schema_path}.slv_AdventureWorks_Calendar")


df_slv_calendar.limit(10).display()
```

###***Customers***

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

Raw_data_path = f"{brz_path}AdventureWorks_Customers"

df_slv_customers = spark.read.format("csv").option("header", True).option("inferSchema", True).load(Raw_data_path)

df_slv_customers=df_slv_customers.withColumn("Full_name",concat(col('Prefix'),lit(""),col("FirstName"),lit(" "),col("LastName"))).withColumn("MaritalStatus",when(col("MaritalStatus")=="M",lit("Married")).otherwise(lit("Single"))).withColumn('Gender',when(col("Gender")=="M",lit("Male")).otherwise(lit("Female"))).withColumn('HomeOwner',when(col("HomeOwner")=="Y",lit("Yes")).otherwise(lit("No"))).withColumn("_source_file", col("_metadata.file_path")).withColumn("injested_at",current_timestamp()).drop(col("Prefix"))

df_slv_customers.write.mode('overwrite').format("delta").option("overwriteSchema", "true").option("path",f"{slv_path}AdventureWorks_Customers").saveAsTable(f"{Silver_schema_path}.slv_AdventureWorks_Customers")


df_slv_customers.limit(10).display()
```

###***Products***

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

Raw_data_path = f"{brz_path}AdventureWorks_Products"

df_slv_Products = spark.read.format("csv").option("header", True).option("inferSchema", True).load(Raw_data_path)

df_slv_Products=df_slv_Products.withColumn('ProductName',split(col("ProductName"),",")[0]).withColumn("ProductSKU",regexp_extract(col("ProductSKU"), r"^([^-]*-[^-]*)", 1)).withColumn("ProductSize",regexp_replace(col("ProductSize"), r"([0-9]+)", "R")).withColumn("ProductStyle",regexp_replace(col("ProductStyle"), r"([0-9]+)", "R")).withColumn("_source_file", col("_metadata.file_path")).withColumn("injested_at",current_timestamp())


df_slv_Products.write.mode('overwrite').format("delta").option("overwriteSchema", "true").option("path",f"{slv_path}AdventureWorks_Products").saveAsTable(f"{Silver_schema_path}.slv_AdventureWorks_Products")


df_slv_Products.limit(10).display()
```
